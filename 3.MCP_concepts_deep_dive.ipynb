### Chapter 3 — MCP Concepts Deep Dive: Tools, Resources, Prompts

This is the most important chapter.

Because MCP is **NOT just tools**.

It introduces **three primitives**:

1. Tools → Actions
2. Resources → Data
3. Prompts → Intelligence templates

These three together make MCP extremely powerful.

---

#### 1️⃣ MCP Tools — Executable Actions

This is closest to what you already know.

Tools are functions the agent can execute.

Examples:

* run SQL query
* send email
* call API
* create file
* deploy code

---

#### Example MCP Tool

```python
def get_weather(city: str) -> str:
    """Get weather of a city"""
    return "32°C"
```

In MCP, this tool is exposed via server.

Agent discovers it dynamically.

---

#### Tool Lifecycle in MCP

Step-by-step:

```text
1. Agent asks server → tools/list
2. Server returns tool schema
3. LLM decides tool to use
4. Agent sends tools/call request
5. Server executes tool
6. Result returned
```

---

#### Tool Schema (important concept)

MCP tools include structured schema:

```json
{
  "name": "get_weather",
  "description": "Get weather of city",
  "inputSchema": {
    "type": "object",
    "properties": {
      "city": { "type": "string" }
    }
  }
}
```

This allows LLM to understand tool automatically.

No manual wiring.

---

#### 2️⃣ MCP Resources — Data Providers (VERY POWERFUL)

Resources are NOT actions.

They are **data sources**.

This is new.

Agent can read them like files.

---

#### Examples of Resources

* Database records
* Files
* Logs
* PDFs
* Codebase
* API data

---

#### Example Resource

```text
resource://customer/123
```

Agent reads it.

Gets structured data.

No function call needed.

---

#### Why resources exist?

Because sometimes agent needs data, not action.

Example:

Instead of tool:

```python
get_customer(id)
```

MCP allows:

```text
read resource://customer/123
```

Cleaner abstraction.

---

#### 3️⃣ MCP Prompts — Reusable Intelligence Units

Prompts are reusable templates.

This is extremely powerful.

Example:

```text
Prompt: code_review

"Review this code and suggest improvements"
```

Instead of rewriting prompt every time.

Agent can request prompt from server.

---

#### Why prompts exist?

Because intelligence can be modular.

Different teams can create expert prompts.

Example:

* security review prompt
* sql optimization prompt
* debugging prompt

Reusable across agents.

---

#### 4️⃣ Difference Between Tools, Resources, Prompts

| Type     | Purpose              | Example              |
| -------- | -------------------- | -------------------- |
| Tool     | Execute action       | send_email()         |
| Resource | Provide data         | customer record      |
| Prompt   | Provide intelligence | code review template |

---

#### 5️⃣ How Agent Uses All Three Together

Example scenario:

User asks:

> "Analyze customer 123 and send summary email"

Agent workflow:

```text
1. Read resource → resource://customer/123
2. Use prompt → summarize_customer
3. Use tool → send_email
```

This is structured intelligence pipeline.

---

#### 6️⃣ Traditional agents only had Tools

Old architecture:

```text
Agent
 └── Tools only
```

MCP architecture:

```text
Agent
 ├── Tools
 ├── Resources
 └── Prompts
```

Much more powerful.

---

#### 7️⃣ Why Resources are Revolutionary

Before MCP, you had to expose everything as tools.

Even reading data required tool.

Example:

```python
def read_file(path: str) -> str:
```

Now MCP supports direct data abstraction.

This makes context sharing cleaner.

---

#### 8️⃣ Capability Discovery

Agent discovers everything dynamically:

```text
tools/list
resources/list
prompts/list
```

Server responds.

Agent updates its capabilities.

Fully dynamic.

---

#### 9️⃣ Internal Flow Diagram

```text
Agent starts
   ↓
Discover tools
Discover resources
Discover prompts
   ↓
User asks question
   ↓
Agent chooses:
  tool OR resource OR prompt
   ↓
Server responds
```

---

#### 🔟 Why this matters for advanced agents

This enables:

* Context-aware agents
* Persistent knowledge
* Shared intelligence layer
* Modular AI systems

Without MCP, this is extremely hard.

Cleaner.

Scalable.

#### Key Insight

Tools = DO something
Resources = READ something
Prompts = THINK something

This is complete intelligence stack.
