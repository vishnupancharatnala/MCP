### Chapter 2 — MCP Architecture and Core Components

Now we move from **why MCP exists** → to **how it is structured internally**.

This is where things become architectural.

---

###  The 3 Main Components of MCP

MCP introduces a clean separation between:

1. **Host**
2. **Client**
3. **Server**

Let’s understand each one.

---

#### 🧠 1. Host (The AI Application)

The **Host** is the application that runs the LLM.

Examples:

* OpenAI products (like ChatGPT)
* Anthropic products (like Claude Desktop)
* Cursor editor
* Your custom Python agent

The host:

* Talks to the LLM
* Decides when to use tools
* Embeds MCP client

Think of host as:

> The main AI application.

---

#### 🔌 2. MCP Client

The **Client** lives inside the Host.

Its job:

* Connect to MCP server
* Discover tools
* Call tools
* Return results to LLM

It translates:

LLM tool call → MCP protocol request

It is like an adapter layer.

---

#### 🖥 3. MCP Server

The **Server** exposes:

* Tools
* Resources
* Prompts

It runs independently.

Can be:

* Local process
* Remote service
* Container
* Cloud server

Multiple agents can connect to same server.

That’s powerful.

---

#### 2️⃣ High-Level Architecture Flow

```
User
  ↓
Host (LLM + Agent)
  ↓
MCP Client
  ↓
MCP Server
  ↓
Tools / Resources
```

Notice something important:

Agent does NOT directly call tools.

It sends structured MCP request.

---

#### 3️⃣ What Makes MCP Different from Normal Tool Calling?

You already know traditional tools:

```python
def send_email(to: str, subject: str): ...
```

These are static.

MCP introduces:

#### Dynamic Capability Discovery

Agent can ask:

```
"What tools do you have?"
```

Server responds with schema.

Agent updates its context.

No hardcoding.

---

#### 4️⃣ Inside MCP Server

An MCP server exposes three things:

| Component | What it does              |
| --------- | ------------------------- |
| Tools     | Executable functions      |
| Resources | Data providers            |
| Prompts   | Reusable prompt templates |

We’ll deep dive next chapter.

---

#### 5️⃣ Multiple Server Architecture

You can connect to multiple MCP servers:

```
Agent
 ├── Filesystem MCP Server
 ├── Database MCP Server
 ├── GitHub MCP Server
 └── Internal Tools MCP Server
```

Each team can build their own server.

Agents plug into them.

No integration nightmare.

---

#### 6️⃣ Real Example Architecture

For example, Claude Desktop connects to:

* Filesystem server
* Git server
* Database server

The LLM doesn’t know these beforehand.

They are discovered at runtime.

That is the power.

---

#### 7️⃣ MCP Communication Style

MCP is JSON-RPC based.

Meaning communication looks like:

```json
{
  "method": "tools/list"
}
```

Server responds:

```json
{
  "tools": [
    {
      "name": "read_file",
      "input_schema": {...}
    }
  ]
}
```

It’s standardized.

Not Python-specific.

Works in:

* Python
* JS
* Go
* Rust

---

#### 8️⃣ Visual Mental Model

Imagine layers:

```
┌────────────────────┐
│        Host        │
│ (LLM + Agent Logic)│
└─────────┬──────────┘
          │
┌─────────▼──────────┐
│      MCP Client    │
└─────────┬──────────┘
          │
┌─────────▼──────────┐
│      MCP Server    │
│  Tools | Resources │
└────────────────────┘
```

Each layer has clear responsibility.

Clean architecture.

---

#### 9️⃣ Why This Matters Architecturally

Without MCP:

Agent owns tools.

With MCP:

Infrastructure owns tools.

This enables:

* Microservice-style AI
* Team-level tool ownership
* Secure sandboxing
* Dynamic extensibility

---

#### 1️⃣0️⃣ Key Insight

MCP transforms:

From: Agent-centric architecture
To: Tool-service architecture

This is similar to how:

Monolith → Microservices

---

#### 🏪 Analogy — Restaurant System

Imagine:

👤 Customer → Host
🧑‍🍳 Kitchen → MCP Server
🧾 Waiter → MCP Client
📖 Menu → Tool list

Flow:

Customer asks waiter
Waiter checks kitchen menu
Kitchen prepares food
Waiter brings result

Customer never goes to kitchen directly.

That’s MCP.
